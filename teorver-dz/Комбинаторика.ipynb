## Задача 1.9
> Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

```python
from math import factorial

n = 7
n1, n2, n3 = 3, 2, 2

result = factorial(n) // (factorial(n1) * factorial(n2) * factorial(n3))

print("Количество семизначных чисел:", result)
```

## Задача 1.10 
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки;
> б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

* Рассмотрим вероятность события $A$ – получится слово ТОР из 3 карточек

```python
from math import factorial, perm

n = 6

a = 1 / perm(n, 3)
b = 1 / factorial(n)

print("а) P(A) = 1 /", perm(n, 3), "=", round(a, 10))
```

* Рассмотрим вероятность события $B$ – получится слово ТЕОРИЯ из 6 карточек

```python
print("б) P(B) = 1 /", factorial(n), "=", round(b, 10))
```

## Задача 1.11
> Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

* Рассмотрим вероятность события $A$ – получится слово АНАНАС из 6 карточек

```python
from math import factorial

n = 6
p3 = factorial(3)
p2 = factorial(2)
p6 = factorial(6)

m = p3 * p2
result = m / p6

print("Вероятность события A =", p3, "*", p2, "/", p6, "=", round(result, 10))
```

## Задача 1.16

>За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом;
>б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

```python
from math import factorial

n = factorial(10)

m1 = 2 * factorial(5) ** 2
a = m1 / n

m2 = factorial(5) * 2 ** 5
b = m2 / n
```

* Рассмотрим вероятность события $A$ – никакие два лица одного пола не сядут рядом

```python
print("а) P(A) =", m1, "/", n, "=", round(a, 5))
```

* Рассмотрим вероятность события $B$ – супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

```python
print("б) P(B) =", m2, "/", n, "=", round(b, 5))
```

## Задача 1.17

> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе;
> б) в семь купе; в) в три купе.

* Рассмотрим вероятность события $A$ – пассажиры попали в два купе

```python
from math import comb

n = comb(36, 7)

m1 = comb(9, 2) * comb(8, 7)
a = m1 / n

m2 = comb(9, 7) * 4 ** 7
b = m2 / n

m3 = comb(9, 3) * (6 * comb(4, 4) * comb(4, 2) * comb(4, 1) +
                   3 * comb(4, 3) * comb(4, 3) * comb(4, 1) +
                   3 * comb(4, 3) * comb(4, 2) * comb(4, 2))
d = m3 / n
```

```python
print("а) P(A) =", m1, "/", n, "=", round(a, 7))
```

* Рассмотрим вероятность события $B$ – пассажиры попали в семь купе

```python
print("б) P(B) =", m2, "/", n, "=", round(b, 5))
```

* Рассмотрим вероятность события $C$ – пассажиры попали в три купе

```python
print("в) P(С) =", m3, "/", n, "=", round(d, 5))
```
